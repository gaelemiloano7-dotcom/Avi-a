# Importar Random Forest

Se importa el algoritmo `RandomForestClassifier` desde la librería Scikit-learn para crear un modelo de bosque aleatorio.

```python
from sklearn.ensemble import RandomForestClassifier
```

# Crear modelo Random Forest

En este bloque se crea el modelo `RandomForestClassifier` configurando diferentes parámetros para generar múltiples árboles de decisión y mejorar la precisión del modelo.

- `n_estimators=100` crea 100 árboles.
- `criterion="gini"` utiliza el índice Gini para dividir datos.
- `max_features="sqrt"` selecciona variables aleatorias.
- `bootstrap=True` permite usar muestras aleatorias con reemplazo.
- `max_samples=2/3` usa dos tercios de los datos por árbol.
- `oob_score=True` calcula la precisión con datos no utilizados durante el entrenamiento.

```python
from sklearn.ensemble import RandomForestClassifier
bosque = RandomForestClassifier(n_estimators=100, criterion="gini", max_features="sqrt", bootstrap=True, max_samples=2/3, oob_score=True)
```

# Entrenar el modelo

En este bloque se entrena el modelo Random Forest utilizando los datos del dataset.  
Las variables de entrada corresponden a todas las columnas excepto `"ingreso"`, mientras que la columna `"ingreso"` se utiliza como variable objetivo.

```python
import pandas as pd
personas = pd.read_csv("ingresos.csv")
bosque.fit(personas [personas.columns[:-1]].values, personas ["ingreso"].values)
```

# Realizar predicciones y evaluar el modelo

En este bloque se realiza una predicción utilizando nuevos datos de ejemplo.  
Después, se calcula la precisión del modelo utilizando todo el dataset y finalmente se muestra el `OOB Score`, que evalúa el rendimiento usando datos que no participaron en el entrenamiento de cada árbol.

```python
print(bosque.predict([[50, 16, 1, 1, 40]]))
print(bosque.score (personas [personas.columns[:-1]].values, personas ["ingreso"].values))
print(bosque.oob_score_)
```

# Importar librerías para gráficas

En este bloque se importan las librerías necesarias para visualizar los árboles de decisión generados por el modelo Random Forest.

```python
import matplotlib.pyplot as plt
from sklearn import tree
```

# Visualizar árboles de decisión

En este bloque se recorren todos los árboles generados por el modelo Random Forest y se muestran gráficamente utilizando `plot_tree`.

Esto permite observar cómo cada árbol toma decisiones a partir de las variables del dataset.

```python
for arbol in bosque.estimators_:
  tree.plot_tree (arbol, feature_names=personas.columns[:-1])
  plt.show()
```
